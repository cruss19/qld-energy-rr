# 09 — TCN ensemble comparison on the frozen 2020 evaluation

This results-only notebook presents the five completed three-seed RR TCN families. All members were fitted on 2015–2019 for durations fixed from the median official best epoch across the 2017, 2018 and 2019 development folds. The frozen 2020 period was scored only after the member states and ensemble rules were fixed. No model is fitted or rescored here.

## Ensemble contract

- Seeds: 42, 142 and 242.
- TCN1, TCN2, TCN3 and TCN_star use an equal-weight arithmetic mean of member point forecasts.
- TCN_starNLL uses the established equal-weight mixture of the three Student-t predictive distributions.
- No ensemble weight was selected from 2020 outcomes.
- The five ensembles use the same 105,402 forecast origins and all six horizons.
- Statistical references are not included here because they predict only the 30-minute endpoint; the common endpoint comparison is Notebook 08.
